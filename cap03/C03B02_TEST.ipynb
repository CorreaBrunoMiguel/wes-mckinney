Q5 - Agrupamento por chave com `setdefault`

```python
# Dada
words = ["ana", "amor", "bola", "banda", "casa"]
```

Crie um dict `g` que agrupa palavras pela primeira letra, usando `setdefault`.

Ex.: `{"a": ["ana", "amor"], "b": ...}

Rubrica: estrutura correta, acumula sem `if` explícito, ordem das inserções preservada (ordem da lista)

```python
g = {}
for word in words:
    g.setdefault(word[0], []).append(word)

print(g)
```

---

Q6 - Mesmo agrupamento com `defaultdict`

Refaça o Q5 usando `collections.defaultdict(list)`.

Rubrica: Uso correto do `defaultdict`, resultado equivalente.

```python
from collections import defaultdict

g = defaultdict(list)
for word in words:
    g[word[0]].append(word)
print(g)  # tipo defaultdict
print(dict(g))  # tipo dict
```

Q10 - Comprehensions (list/set/dict)

```python
# dada
nums = [1, 2, 3, 4, 5, 6]
```

(a) List comprehensions com apenas pares ao quadrado

```python
[n**2 for n in nums if n % 2 == 0]
```

(b) Set comprehensions com os restos de `n % 3`.

```python
{n % 3 for n in nums}
```

(c) Dict comprehensions mapeando `n -> n*n` apenas para ímpares:

```python
{n: n * n for n in nums if n % 2 != 0}
```

Q11 - Flattern com filtro (aninhado):

(b) Crie `flat_even` contendo apenas os pares.

```python
flat_even = [num for lista in matrix for num in lista if num % 2 == 0]
flat_even
```
