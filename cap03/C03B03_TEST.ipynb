```python
# Q11)
names = ["Bob", "anna", "Albert", "Bruno", "alex", "Bea", "carla"]

# 1)
first_letter = lambda x: x[0]
for letter, name in itertools.groupby(names, key=first_letter):
    print(letter, list(name))
```

```python
# 2)
for letter, name in itertools.groupby(
    sorted(names, key=first_letter), key=first_letter
):
    print(letter, list(name))
```

Q11) Explicação

No primeiro resultado ficou quebrado com chaves repetidas porque itertools.groupby percorre os elementos numa ordem consecutiva, sendo assimo ideal seria passar o iterável já na ordem "correta"

---

```python
# Q12)
def safe_ints(s):
    try:
        int(s)
    except ValueError:
        return None
    else:
        return int(s)
    finally:
        print(f"String recebida: {s}")


safe_ints("olá")
```

Q12) - Explicação

ValueError é mais direto neste contexto, Exception é muito abrangente pode mascarar erros e bugs mais profundos.
