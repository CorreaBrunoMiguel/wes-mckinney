## C03B04 - Respostas Descritivas

1. É uma conexão entre Python e um arquivo do sistema operacional, possibilitando a manipulação do arquivo com métodos especificos (read, write, etc). Quando trabalhamos com arquivo e abrimos a conexão precisamos fechar a mesma pois além do consumo de memória podem gerar erros e bugs.

2. Quando usamos `with open(path)as f:`, garantimos que iremos abrir o arquivo e fechá-lo após a lógica de manipulação do mesmo. É um modo "seguro" de trabalhar com arquivos locais.

3. Para cada modo temos:

- `r` -> somente leitura, se não existir gera erro se existir abre para leitura
- `w` -> modo de escrita, sobreescreve arquivo se já existir, caso contrário ele cria
- `x` -> cria um arquivo novo mas somente se não houver o arquivo, ou seja, se já existir ele gera uma falha
- `a` -> concatena a um arquivo existente, caso não exista ele cria um novo

4. `r` é apenas leitura e trata o arquivo como texto, já `rb` trabalha com bytes podendo ser de 10 até 40; str seria texto puro já bytes trabalhará com encoding retornando texto em bytes;

5. `read()` lê o arquivo inteiro, ideal para pequenos arquivos, arquivos grandes podem influencar no desempenho e memória;

`read(n)` lê n caracteres (texto ou bytes), ideal para "quebrar" arquivo em pequens trechos, economiza memória pois faz conexão com pequenos trechos

`readline()` lê apenas uma linha devolvendo "\n" (quebra de linha) se houver

`readlines()` cria uma lista de linhas do arquivo, consome muita memória

`for line in f:` melhor forma para iterar oir um arquivo pois entrega linha a linha;

6. `write()` e `writelines()` por padrão não incluem '\n'; Não sei dizer com certeza o que pode causar a falta de inclusão do '\n', mas acho que arquivos como .csv pode causar problemas na leitura;

7. `seek()` move ponteiro para posição especificada, já `tell()` devolve a posição atual do arquivo como inteiro; Não sei porque em binário é mais correto, mas deve ter ligação com encoding e decoding, porque caracteres especiais como " " e outros podem gerar erros;

8. `0` inicio do arquivo, `1` posição atual, `2` final do arquivo, `whence` ajuda a reposicionar a previsibilidade;

9. Principalmente quanto a remover ou não espaços de final de linhas, `strip()` tende a remover todos, o que pode não ser o ideal, pois não sabemos se é ou não importante para o arquivo naquele momento.

10. `FileNotFoundError` - quando tentamos ler um arquivo num path onde não existe;
`PermissionError` - quando tentamos ler/escrever um arquivo que não permite tais ações
`UnicodeDecodeError` - há erro no encoding para modo textual

11. Para que não ocorra suposições ao arquivo, devemos explicitamente demonstar arquivos (paths), evitando dependências do "cwd"

## C03B04 — Resumo pré-avaliação (para B6)

### 1) `open()` e modos (o que acontece com arquivo existente/inexistente)
- `r`  : leitura; erro se não existir (`FileNotFoundError`)
- `w`  : escrita; cria se não existir; SE existir, TRUNCA (zera)
- `x`  : cria exclusivo; erro se já existir (`FileExistsError`)
- `a`  : append; cria se não existir; escreve sempre no fim
- `r+` : leitura/escrita; não trunca; cursor no início
- `w+` : leitura/escrita; TRUNCA (zera)
- `a+` : leitura/escrita; escrita no fim (cursor pode variar)

Texto vs binário:
- texto (`"r"`, `"w"`) -> trabalha com `str` e usa `encoding`
- binário (`"rb"`, `"wb"`) -> trabalha com `bytes` e NÃO usa encoding

## 2) Leitura: qual usar e por quê
- `read()`        -> tudo (ok só para arquivo pequeno)
- `read(n)`       -> chunk (n chars em texto; n bytes em binário)
- `readline()`    -> 1 linha
- `readlines()`   -> lista de linhas (cuidado com memória)
- `for line in f` -> melhor padrão: streaming linha a linha, eficiente

Limpeza de linha:
- `rstrip("\n")`  -> remove só newline (mantém espaços finais)
- `rstrip()`      -> remove whitespace do fim (pode remover espaços significativos)
- `strip()`       -> remove whitespace início e fim (perigoso se indentação/espaços importam)

## 3) Escrita: o detalhe que mais derruba
- `write()` e `writelines()` NÃO adicionam `\n`
  => se você quer “uma linha por registro”, adicione `"\n"` manualmente
  ou use `print(..., file=f)` (newline por padrão).

Exemplo mental: se você fizer `write("a")` depois `write("b")`,
o arquivo vira "ab" (não "a\nb\n").

## 4) `with` e robustez (regra de ouro)
Use SEMPRE:
`with open(path, mode, encoding="utf-8") as f: ...`

Motivos:
- fecha mesmo com exceção
- garante flush em escrita
- evita vazamento de descritor ("too many open files")

## 5) `tell()` e `seek()` (precisão)
- Em binário (`rb`): `tell()` ~ offset em bytes previsível; `seek()` reposiciona por bytes.
- Em texto (`r`): `tell()` pode ser “opaco” por buffering/decoding/newline translation.
  => para discutir offset em bytes com precisão, demonstre em `rb`.

`seek(offset, whence)`:
- `whence=0`: a partir do início
- `whence=1`: a partir da posição atual
- `whence=2`: a partir do fim

## 6) Exceções comuns (saiba identificar)
- `FileNotFoundError`: abrir em `r` arquivo inexistente
- `PermissionError`: sem permissão (path protegido/sem acesso)
- `UnicodeDecodeError`: encoding incorreto ao ler texto
- `FileExistsError`: abrir com `x` quando já existe
