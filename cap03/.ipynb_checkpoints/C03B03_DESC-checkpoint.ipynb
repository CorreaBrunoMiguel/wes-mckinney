## C03B03 — Descritivas

1) Explique, com suas palavras, como o Python resolve nomes (variáveis/funções) quando um identificador é usado dentro de uma função. Descreva a lógica de escopo (local vs global) e o impacto prático disso em código real.

2) O que `global` faz dentro de uma função? Por que, em geral, é considerado um antipadrão? Dê pelo menos duas alternativas mais seguras para lidar com “estado” sem recorrer a `global`.

3) Em Python, o que significa “retornar múltiplos valores”? Explique o que de fato é retornado (do ponto de vista do tipo/estrutura) e como o desempacotamento funciona na atribuição.

4) Defina “funções são objetos de primeira classe” e cite três consequências práticas (ex.: passar função como argumento, armazenar em estruturas, retornar função etc.). Dê exemplos de casos reais onde isso é útil.

5) Quando usar `lambda` é adequado e quando vira prejuízo de legibilidade/manutenção? Compare `lambda` com `def` usando critérios objetivos (tamanho, reutilização, debug, nomeação).

6) Explique o que é aplicação parcial (partial application) e qual problema ela resolve. Em que situações `functools.partial` é mais apropriado do que criar uma função “wrapper” manualmente?

7) Explique o que é um gerador e como `yield` muda o comportamento de uma função. Compare geradores com listas em termos de: memória, desempenho, e fluxo de execução (pausa/retomada).

8) O que é “exaustão de iterador”? Dê um exemplo conceitual (sem precisar colar código) de um iterador que é consumido uma vez e depois “some”. Como corrigir quando você precisa reutilizar os dados?

9) Descreva o “protocolo de iteração” do Python: o que significa um objeto ser iterável vs ser iterador? Qual é o papel de `iter()` e `next()` nessa história?

11) Explique a estrutura completa `try / except / else / finally`. Em que situações usar `else` melhora o design do código? Por que capturar exceções específicas é melhor do que `except Exception` em código de domínio?


## C03B03 - Respostas

1. Python usa o conceito de escopo, quando um identificaor é usado dentro de uma função, ele "existe" apenas enquant esta mesma função existe; Ao rodar uma função o identificador é criado, usado, analisado etc, se a função não retornar o valor desse identificador o mesmo morre após função finalizar. Identificadores de escopo global estão acessíveis a todo o código, dentro de funções, condicionais, qualquer bloco de código acessível, já de escopo local possui limitação ao bloco que estiver inserido, ou em blocos aninhados;

2. Usar `global` dentro de uma função é contraprodutivo, pois a mesma estará acessível em todo local, podendo sofrer reatribuição, entre outros que possam comprometer o código. Se você quiser usar a variável de dentro da função fora, você pode retornar a mesma ou atribuí-la fora do escopo de função;

3. Em python retornar multiplos valores, significa retornar uma tupla. Logo se quisermos desempacotar uma função (seu retorno) podemos usar o método geral de desempacotamento de tuplas, usando variaveis separadas por vírgula na mesma linha, recebendo a função; É importante saber qual tamanho da tupla que será retornada;

4. Porque podem ser pasadas como argumentos, armazenadas em lists e dicts, podem ser atribuídas a variáveis, ser retornadas por outras funções;

5. `lambdas`são ideais para funções que possuem pouca lógica, que sejam mais diretas, ideais para serem passadas como argumentos e também como retorno de outras funções. Se a função possui uma lógica interna masi avnaçada, com 2+ linhas de código, ou seja, for mais robusta o ideal é usar `def`, assim como a necessidade de ser nomeada seja uqal for o motivo deve ser usado `def` pois `lambda` são funções anônimas, `def` também é ideal quano a função for reutilizada e facilida debugs;

6. É quando precisamos fixar o valor de um argumento, tornando ela mais direta naquele sentido da variável fixada; Ideal para paginar respostas, limitar retorno entre outros; É como se você pré-configurace o retorna daquela função; Ela é mais direta do quê um `wrapper` pois a mesma pode ser pré-configurada de várias formas, já `wrapper`é definida para um único e necessário propósito;

7. Geradores são funções que ao invés de retornar um valor "direto", gera um `generator` uma sequência dormente que deverá ser ativada de alguma forma (list, for, etc) para que os valores "apareçam"; Usam menos memória que list pois o que existe é apenas uma "informação" daquilo que ela contém ao invés de valores de fato, após serem usados os geradores desaparecem;

8. Significa dizer que após ser usado o iterador desaparece, seu namespace é consumido; se houver necessidade de mais uso podemos criar laços ou reatribuir, se houver necessidade de repetição poemos até usar `seed` entre outros;

9. iterador é o tipo já criado específico, já iterável é que pode ser "percorrido", quando usamos `iter()` retornamos a sequência presente já `next()` percorre os valores um a um, dando um maior controle ao user;

10. `itertools.groupby` percorre uma sequência e organiza os valores por chaves iguais (respeitando a callback passada), porém ele é aplicado em elementos consecutivos, e se os dados não estiverem ordenados ele irá criar chaves repetidas; o Ideal é aplicar em sequências orgnaizadas, sorteadas pela parametro que você escolher;

11. `try` irá "tentar" aplicar a lógica criada; `except` irá estourar se a lógica do `try` houver "erros"/exceções; `else` é uma boa prática para mostrar, aplicar, etc logicas quando o código não entrar em `except`; `finally` irá rodar independente de qualquer coisa, ideal para fechamentos, logs, limpezas, códigos que precisam ser rodados; `except Exception` pode mascarar erros, bugs entre outros pois ele é mais amplo e esconde a causa real;

   Correto: aumenta acoplamento e risco de efeitos colaterais. Faltou dizer objetivamente que `global` faz o nome referenciar o namespace global (permitindo atribuição nele). Alternativas: retornar valor; encapsular estado (classe); closure + `nonlocal` quando há função aninhada.

   Ideia central correta (“fixar argumento”). Exemplos ficaram genéricos (paginação/limitar retorno não é o caso típico). O ponto forte é “adaptar assinatura” e pré-configurar funções para APIs que chamam com menos parâmetros (ex.: `key=`, `map`).

8. Exaustão de iterador
   Aqui ficou fraco: “namespace consumido” e “seed” não se aplicam. O correto: iterador mantém estado de posição; ao consumir com `list()`/`for`, ele avança até acabar. Correções: recriar o iterador, materializar em lista uma vez, ou usar `itertools.tee` quando fizer sentido.

9. Iterável vs iterador (`iter()`/`next()`)
   Você acertou a intuição. Para precisão: iterável implementa `__iter__` (retorna um iterador); iterador implementa `__next__` (e normalmente `__iter__` retornando ele mesmo).

    Muito bom: agrupamento por consecutivos + necessidade de ordenar pela mesma chave antes para agrupamento consistente.

    Cobriu bem o papel do `else` e do `finally` e o risco de `except Exception`. Ajuste: `except` não “estoura”, ele trata; e o ideal é capturar tipos específicos no domínio.
