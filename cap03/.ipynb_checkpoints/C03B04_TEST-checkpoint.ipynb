## C03B04 - Respostas Avaliação de Bloco

Q1) 
- `r` -> se o arquive não existir retorna um erro: `FileNotFoundError`, se houver abre uma conexão para leitura
- `w` -> se o arquivo já existir ele sobrescreve o mesmo, se não ele cria um novo arquivo, na verdade ele TRUNCA o arquivo antigo (Zera)
- `x` -> se já existir gera o erro: `FileExistsError`, se não existir cria um novo
- `a` -> se existir adiciona ao final do mesmo, se não cria um novo

---

Q2) Se abrirmos um arquivo usando método `r` ele trata como `str`, Python usa o enconding para mostrar  arquivo, Já se abrirmos com `rb` ele "devolve" o arquivo em `bytes`, onde não há necessidade de enconding;

---

```python
# Q3)
def count_lines(path):
    lines = 0
    with open(path) as f:
        for line in f:
            lines += 1
    return lines


count_lines(segismundo)
```

---

```python
# Q4)
with open("data/out.txt", "w") as f:
    f.write("A\n")
    f.write("B\n")
    f.write("C")
```

---

Q5) 
- `strip()`: remove espaço/s em branco, no inicio e final da linha se houver, (perigoso pode gerar perda de dados)
- `rstrip()`: remove espaço/s no final da linha (também pode remover espaços sgnificativos)
- `rstrip("\n")`: remove apenas _newline_ mantendo os espaços

---

Q6)Usando `with` teremos a certeza que o arquivo será fechado, independente se haja error ou não, já manualmente se houver geração de erro no caminho Python pode nunca interpretar o fechamento do arquivo

---

```python
# Q7)
with open("data/segismundo.txt") as f:
    lines = f.read(30)
    print(lines)
```

```python
with open("data/segismundo.txt") as f:
    lines = f.read()
    print(len(lines))
```

---

```python
# Q8)
with open("data/segismundo.txt", "rb") as f:
    f.read(10)
    print(f.tell())
    f.seek(0)
    print(f.read(10))
```

Porque quando abrimos com binário temos a certeza de onde o ponteiro do arquivo se encontra 1 byte 1 posição;

---

Q9) 
- `FileNotFoundError`: `with open(path) as f:` quando `path` não existe
- `PermissionError`: tentar abrir conexão com um arquivo quando o path está protegido ou sem acesso;
- `FileExistsError`: modo `x` do `open()` exige esxclusividade, esse erro ocorre quando tentamos abrir arquivo que já existe
