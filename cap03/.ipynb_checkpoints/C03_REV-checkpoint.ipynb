# C03 - Revisão

## Parte 1 — Estruturas de dados e sequências (C03B01 + C03B02)

1. Tupla vs lista (imutabilidade e efeito prático)

* tuple: imutável; boa para “registro” (dados que não deveriam mudar), pode ser chave de dict/set se todos os elementos forem hashable.
* list: mutável; boa para coleção que cresce/encolhe.

Armadilhas recorrentes:

* Tupla de 1 elemento precisa de vírgula: `(1,)`.
* Uma tupla pode conter uma lista mutável; a tupla continua imutável, mas a lista interna muda (isso surpreende).

2. Unpacking (atribuição múltipla)

* Base: `a, b = (1, 2)`; swap: `a, b = b, a`.
* Com excedente: `head, *rest = seq`.
* Em loop: `for k, v in d.items(): ...`

Onde isso cai em prova: manipular pares (zip/items), extrair “primeiro e resto”, trocar variáveis sem temporário.

3. Slicing e suas 3 formas (importante lembrar o que copia vs altera)

* `seq[a:b:c]` gera nova sequência (cópia) na maioria dos casos.
* Atribuição por slice em lista altera a lista: `lst[2:4] = [x, y, z]` (pode mudar tamanho).
* Reverso: `seq[::-1]` (cópia/review); `lst.reverse()` (in-place); `reversed(seq)` (iterador).

4. sorted vs sort e key

* `sorted(iterable, key=..., reverse=...)` retorna nova lista.
* `list.sort(...)` altera a lista.
* Empate: key composta (tupla): `key=lambda x: (x[1], x[0])`.

5. Dict (mapa) e padrões que você precisa dominar sem pensar

* Acesso seguro: `d.get(k, default)` evita `KeyError`.
* Remoção segura: `d.pop(k, default)` evita `KeyError`.
* Iteração correta: `for k in d`, `for v in d.values()`, `for k, v in d.items()`.

Agregação por chave:

* `setdefault`: “se não existe, cria” → `d.setdefault(k, []).append(v)`
* `defaultdict(list)`: simplifica o append, mas cria a chave ao acessar (efeito colateral real).

6. Set (conjunto) e por que ele é diferente de list

* Membership (`in`) no set é O(1) médio; em list é O(n).
* Operações: união `|`, interseção `&`, diferença `-`, dif. simétrica `^`.
* Elementos precisam ser hashable: `list`/`dict`/`set` não entram; `tuple` entra se todos os itens forem hashable.

7. Comprehensions (poder + limite de legibilidade)

* List: `[f(x) for x in it if cond]`
* Dict: `{k: v for ...}`
* Set: `{x for ...}`
  Regra prática: se virou “árvore de if/for”, migre para `for` explícito.

## Parte 2 — Funções, iteradores/geradores, itertools, exceções (C03B03)

1. LEGB (resolução de nomes) e shadowing

* Local → Enclosing → Global → Builtins.
* Shadowing: um nome local “sombra” um global (o local ganha).
* `global` só quando você quer atribuir no escopo global (evitar em domínio). Alternativas: retornar valor, encapsular em classe, closure com `nonlocal`.

2. Funções são objetos (isso cai muito)

* Você passa função para `key=`, `map`, callbacks, pipelines.
* `sorted(data, key=func)` e desempate com tuplas.
* “Função de ordem superior”: recebe/retorna função.

3. lambda vs def (critério objetivo)

* lambda: expressão curta, uso local (principalmente `key=`).
* def: qualquer lógica que você precisa nomear, reutilizar, debugar, testar.

4. partial (functools.partial)

* Serve para “fixar” argumentos e adaptar assinatura.
  Exemplo mental: você tem `power(base, exp)` e cria `square = partial(power, exp=2)`.

5. Iterável vs iterador e exaustão (ponto crítico)

* Iterável: tem `__iter__` e produz um iterador.
* Iterador: tem `__next__` e mantém estado; quando esgota, acabou.
* Exaustão: consumir um iterador uma vez (for/list/sum) o deixa vazio depois. Se precisa de novo: recrie o iterador ou materialize em lista uma vez.

6. Geradores (`yield`)

* Gerador é um iterador criado por `yield`.
* Vantagem: lazy evaluation (memória) + pipelines.
* Consumo: `for`, `list(...)`, `sum(...)`, etc.

7. itertools e o erro clássico do groupby

* `groupby` agrupa apenas itens consecutivos com a mesma chave.
* Para “agrupar de verdade”, ordene antes pela mesma chave usada no groupby.

8. try/except/else/finally (estrutura completa)

* `except` deve ser específico (ex.: `ValueError`).
* `else` roda só se não houve exceção (ajuda a manter o try pequeno).
* `finally` roda sempre (limpeza/registro).

## Parte 3 — Arquivos e I/O (C03B04)

1. open e modos (o “mapa mental”)

* `r` leitura (não existe → erro)
* `w` escrita (existe → trunca; não existe → cria)
* `x` cria exclusivo (já existe → erro)
* `a` append (não existe → cria; escreve no fim)
* variações com `+`: leitura+escrita

2. Texto vs binário (str vs bytes)

* Texto (`"r"`): `str` + encoding (recomendação: `utf-8` explícito).
* Binário (`"rb"`): `bytes` sem encoding/decoding.

3. Leitura e escrita (detalhe que derruba)

* `write()`/`writelines()` NÃO adicionam `\n`.
* Leitura eficiente: `for line in f`.

4. seek/tell

* Binário: previsível em bytes.
* Texto: pode ser “opaco” por buffering/decoding.

5. Exceções comuns

* `FileNotFoundError`, `PermissionError`, `UnicodeDecodeError`, `FileExistsError`.

## Parte 4 — Como tudo se junta num pipeline típico de capítulo

Você pode pensar assim: arquivos → linhas (iterável) → limpeza (funções/comprehensions) → estrutura (dict/set/list) → agregação (defaultdict/groupby ordenado) → saída (arquivo) com tratamento de exceção.
