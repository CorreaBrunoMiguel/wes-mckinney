### dict

`dict` provavvelmente é a estrutura de dados embutida mais importante de Python. Também chamado de _hash mao_ ou _array associativo_. Consiste numa coleção de pares _chave-valor_ de tamanho flexível, em que a _chave_ e o _valor_ são objetos Python. A forma mais simples de criar um dicionário é usando chaves `{}` e dois pontos para separar as chaves e os valores.

````python
empty_dict = {}
type(empty_dict)
````

````python
d1 = {"a": "some value", "b": [1, 2, 3, 4]}
d1
````

Podemos acessar, inserir ou definir elementos usando a mesma sintaxe utilizada para acessar elementos de uma lista ou de uma tupla:

````python
d1[7] = "an integer"
d1
````

````python
d1["b"]
````

Verificamos se um dicionário contém uma chave usando a mesma sintaxe utilizada para verificar se uma lista ou tupla contém um valor:

````python
"b" in d1
````

É possível apagar valores usando a palavra reservada `del` ou com o método `pop`:

````python
d1[5] = "some value"
d1
````

````python
d1["dummy"] = "another value"
d1
````

````python
del d1[5]
d1
````

````python
ret = d1.pop("dummy")
ret
````

````python
d1
````

Os métodos `key` e `values` oferecem iteradores paras as chaves e valores do dicionário, respectivamente. EMbora os pares chave-valor não estejam em nenhuma ordem em particular, essas funções devolve as chaves e os valores na mesma ordem:

````python
list(d1.keys())
````

````python
list(d1.values())
````

Podemos combinar um dicionário com outro usando o método `update`:

````python
d1.update({"b": "foo", "c": 12})
d1
````

O método `update` altera os dicionários in-place, portanto qualquer chave existente nos dados passados para `update` terão seus valores antigos descartados.

#### Criando dicionários a partir de sequências

```python
mapping = {}

for key, value in zip(key_list, value_list):
    mapping[key] = value
```

Considerando que um dicinário é essencialmente uma coleção de tuplas de 2, a função `dict` aceita uma lista de tuplas de 2:

````python
mapping = dict(zip(range(5), reversed(range(5))))
mapping
````

#### Valores default

```python
if key in some_dict:
    value = some_dict[key]
else:
    value = deault_value
```

Logo, métodos de dicionário `get` e `pop` podem aceitar um valor default a ser devolvido, de mdo que é possível escrever o bloco `if-else` anterior simplesmente como:

```python
value = some_dict.get(key, default_value)
```

Por padrão, `get` devolverá `None` se a chave não estiver presente, enquanto `pop` lançará uma exceção. Com valores a serem _definidos_, um caso comum é aqule em que os valores de um dicionário são outras coleções, como listas. Por exemplo, poderíamos imaginar a classificação de uma lista de palavras de acordo com suas primeiras letras como um dicionário de listas:

````python
words = ["apple", "bat", "bar", "atom", "book"]
by_letter = {}

for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)

by_letter
````

O método de dicionáio `detdefaut` serve exatamente para essa finalidade. O laço for anterior pode ser reescrito assim:

```python
for word in words:
    letter = word[0]
    by_letter.setdefault(letter, []).append(word)
```

O módulo embutido `collections` tem uma classe útil, `defaultdict`, que torna isso mais fácil ainda. Para criar um dicionário, passe um tipo ou uma função para gerar o valor default de cada posição no dicionário:

```python
from collections import defaultdict

by_letter = defaultdict(list)
for word in words:
    by_letter[word[0]].append(word)
```

#### Tipos de chave válidos para dicionários

Enquanto os valores de um dicionário podem ser qualquer objeto Python, as chaves geralmente têm que ser objetos imutáveis, como tipos escalares (int, float, string) ou tuplas (todos os objetos na tupla devem ser imutáveis também). O termo técnico, nesse caso, é _hashability_ (possibilidade de hashing). É possível verificar se um objeto é hashable (pode ser usado como chave em um dicionário) usando a função `hash`:

````python
hash("string")
````

````python
hash((1, 2, (2, 3)))
````

````python
try:
    hash((1, 2, [2, 3]))
except Exception as e:
    print("Error:", e)
````

### set

Um conjunto é uma coleção não ordenada de elementos únicos. Podemos pensar neles como dicionários, mas somente com as chaves, sem os valores. Um conjunto pode ser criado de duas maneiras: com a função `set` ou por meio de um _conjunto literal_, com chaves:

````python
set([2, 2, 2, 1, 3, 3, 3])
````

````python
{2, 2, 3, 3, 3, 1, 4, 5, 5, 4}
````

Conjuntos aceitam _operações matemáticas de conjunto_, como união, intersecção, diferença e diferença simétrica. Considere os dois exemplos de conjuntos a seguir:

````python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7, 8}
````

A união desses dois conjuntos é o conjunto de elementos distintos presentes em qualquer conjunto. Esse conjunto pode ser calculado com o método `union` ou com o operador binário `|`

````python
a.union(b)
````

````python
a | b
````

A intersecção contém os elementos presentes nos dois conjuntos. O operador `&` ou o método `intersection` podem ser usados:

````python
a.intersection(b)
````

````python
a & b
````

Abaixo temos uma tabela que contém uma lista de métodos de conjunto comumente utilizados.

| __Função__ | __Sintaxe alternativa__ | __Descrição__ |
|------------|-------------------------|---------------|
| a.add(x) | N/A | Adiciona o elemento x ao conjunto a |
| a.clear(x) | N/A | Reinicia o conjunto __a__, deixando-o em um estado vazio, descartando todos os seus elementos |
| a.remove(x) | N/A | Remove o elemento x do conjunto a |
| a.pop() | N/A | Remove um elemento arbitrário do conjunto a, gerando um __KeyError__ se o conjunto estiver vazio |
| a.union(b) | a \| b | Todos os elemento únicos de a e b |
| a.update(b) | a\|= b | Define o conteúdo de a como a união dos elementos em a e b |
| a.intersection(b) | a & b | Todos os elementos _tanto_ em a _quanto_ em b |
| a.intersectio_update(b) | a &= b | Define o conteúdo  de a como a interssecção dos elementos de a em b |
| a.difference(b) | a - b | Os elementos em a que não estão em b |
| a .difference_update(b) | a -= b | Define a com os elementos que estão e a, mas que não estão em b |
| a.symmetric_difference(b) | a ^ b | Todos os elementos que estão em a ou em b, mas _não em ambos_ |
| a.symmetric_difference_update(b) | a ^= b | Define a para que contenha os elementos que estão em a ou em b, mas _não em ambos_ |
| a.issubset(b) | N/A | True se os elementos de a estiverem todos contidos em b |
| a.issuperset(b) | N/A | True se os elementos de b estiverem contidos em a |
| a.isdisjoint(b) | N/A | True se a e b não tiverem elementos em comum |

Todas as operações lógicas de conjunto têm contrapartidas in-place, o que permite a você substituir o conteúdo do conjunto à esquerda da operação com o resultado. Para conjuntos muito grandes, isso talvez seja eficiente:

````python
c = a.copy()

c |= b

c
````

````python
d = a.copy()
d &= b
d
````

Para termos elementos do tipo lista em um conjunto, devemos fazer a conversão para uma tupla:

````python
my_data = [1, 2, 3, 4]
my_set = {tuple(my_data)}
my_set
````

Podemos verificar se um conjunto é um subconjunto ou é um superconjunto de outro conjunto:

````python
a_set = {1, 2, 3, 4, 5}
{1, 2, 3}.issubset(a_set)
````

````python
a_set.issuperset({1, 2, 3})
````

Os conjuntos serão iguais se, e somente se, seus conteúdos forem iguais:

{1, 2, 3} == {3, 2, 1}

### List, set e dict comprehensions

_List comprehensions_ são um dos recursos mais amados da linguagem Python. Elas permitem que você componha uma nova lista de modo conciso, filtrando os elementos de uma coleção, transformando os elementos ao passar o filtro, com uma expressão concisa. As _list comprehensions_ assumem o seguinte formato básico: 

`[expr for val in collection if condition]`

Isso aqui é equivalente ao laço `for` a seguir:

```python
result = []
for val in collection:
    if condition:
        result.append(expr)
```

A condição de filtro pode ser omitida, resultando somente a expressão. Por exemplo, dada uma lista de strings, poderíamos filtrar, eliminando aquelas de tamanho 2 ou menores, e também converter as strings para letras maiúsculas, assim:

````python
strings = ["a", "as", "bat", "car", "dove", "python", "bruno"]

[x.upper() for x in strings if len(x) > 2]
````

Set e dict comprenhesions são ua extensão natural, gerando conjuntos e dicionários de modo semelhante e idiomático, em vez de listas. Um _dict comprehension_ tem o seguinte aspecto:

`dict_comp = {expr-chave: expr-valor for value in collection if condition}`

Uma _set comprehension_ se assemelha a uma list comprehension equivalente, exceto pelas chaves no lugar dos colchetes:

`set_comp = {expr for value in collection if condition}`

Assim como as _list comrehensions_, as `set` e `dict` comprehensions são, em sua maior parte, conveniências, mas, de modosemelhante, podem deixar o código mais fácil tanto para escrever qunato para ler. 

````python
unique_lengths = {len(x) for x in strings}
unique_lengths
````

Daria para fazer com `map` mas ainda não estudamos (estudaremos mais a frente)

Como um exemplo simples de _dict comprehension_, poderíamos criar um mapa de consulta dessas strings para seus locais na lista:

````python
loc_mapping = {val: index for index, val in enumerate(strings)}
loc_mapping
````

#### List comprehensions aninhadas

Suponha que tenhamos uma lista de listas constendo alguns nomes em inglês e em espanhol.

````python
all_data = [
    ["John", "Emily", "Michael", "Mary", "Steven"],
    ["Maria", "Juan", "Javier", "Natalia", "Pillar"],
]
````

Você podería ter obtido esses nomes de dois arquivos e decidiu organiza-los por idioma. Suponha agora que quiséssemos obter uma única lista contendo todos os nomes com dois ou mais _e_'s. Certamente poderíamos fazer isso com um laço `for` simples:

```python
name_of_interest = []
for names in all_data:
    enough_es = [name for name in all_data if name.count('e') >= 2]
    name_of_interest.extend(enough_es)
```

Na verdade, podemos encapsular toda essa operação em uma única _list comprehension aninhada_, que terá o seguinte aspecto:

````python
result = [name for names in all_data for name in names if name.count("e") >= 2]
result
````

A primeira vista é um pouco difícil de compreender. As partes com `for` da _list comprhension_ são organizadas de acordo com a orem em que estão aninhadas, e qualquer condição de filtragem será colocada no final, como antes. Eis um outro exemplo em que "linearizamos" uma lista de tuplas de inteiros, convertendo-a em uma lista simples:

````python
some_tuples = [(1,2,3),(4,5,6),(7,8,9)]
````
