1) Compare `d[key]` e `d.get(key)` em termos de comportamento quando a chave não existe. Em quais situações você prefere cada um?

R: `d['key']` acusa um KeyRrror quando o dicionário não possui a chave especificada, já `d.get('key')` retorna um NoneType em caso de chave inexistente. Acho que o uso para cada um está nesta idéia se quero dar erro para mostrar que não existe ou continuar caso não exista;

2) Explique o que update faz “in-place” e qual o risco/efeito colateral quando há chaves repetidas.

R: Ele altera o dicionário in-place, adicionando novas chaves ou sobreescrevendo chaves repetidas, ou seja, no caso `d.update(c)`, se d e c possuírem chaves repetidas a chave de c irá sobrescrever a chave de d;

---

3) Você tem duas listas keys e values. Explique duas formas de construir um dicionário e cite uma armadilha envolvendo tamanhos diferentes.

R: 
```python
x = [1, 2, 3, 4]
y = ["aba", "bar", "ceu", "dia", "ela"]

d = {}
for key, value in zip(x, y):
    d[key] = value

# Ou
for num in x:
    d[num] = y[num]
```

Nos dois métodos o tamanho final do dicionário (len) é ditado pela sequência de menor tamanho, daria para contornar isso utilizando métodos de valores default, obrigando a todos elementos de todas listas estarem incluso

4) Em que cenário setdefault resolve um padrão comum? Dê um exemplo conceitual (pode ser “palavras por letra”).

R: setdefault resolve o KeyError onde se o valor da chave não está detreminado ele adiciona um valor default; Tenho um dicionário do tipo `bruno = {"nome": "Bruno","age": 40}` se eu fizer `x = bruno.setdefault("age") = 30` e `print(x)` resultado será ainda 40 pois já havia, agora se tivermos `y = bruno.setdefault('cidade': 'GCR')` e fizer `print(y)` retornará 'GCR'

8) Explique, com exemplos conceituais, a diferença entre união, interseção e diferença de conjuntos.

R: União é pegar todos os elementos de dois ou mais conjuntos e juntar em um outro conjunto, valores iguais tornam apenas um elemnto (set não repete elementos), interseção é dados os conjuntos a e b, dizemos `a & b` o conjunto formado pelos elementos presentes em a e b já diferença seriam os elementos que estão em a mas não estão em b;
