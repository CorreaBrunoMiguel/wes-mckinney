Q4 (8pts) - Swap e legebilidade

Escreva uma função `swap_if_greater(a, b)` que retorna `(a, b)` em ordem crescente usando apenas desempacotamento (sem variável temporária). (8)

```python
def swap_if_greater(a, b):
    return (a, b) if a <= b else (b, a)
```
