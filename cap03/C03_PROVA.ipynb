```python
# Q10)
vals = [1, 2, 3]

it = iter(vals)

print(f"Primeiro consumo de it: {list(it)}")
print(f"Segundo consumo de it: {list(it)}")
```

Isso acontece porque um iterador é consumido quando usado, ou seja, ele limpa a referência; Um jeito simples é fazer `it = iter(vals)` após primeir consumo, ou seja, recriar o iterador;

```python
# Q11)
def evens(n):
    for x in range(n + 1):
        if x % 2 == 0:
            yield x


list(evens(12))
```

```python
# Q12)
names = ["Bob", "anna", "Albert", "Bruno", "alex", "Bea", "carla"]

names_ordenado = sorted(names, key=str)
names_minusculo = [x.lower() for x in names_ordenado]
names_minusculo
```

```python
for k, group in groupby(names_minusculo, key=lambda x: x[0]):
    print(k, list(group))
```

Se não houver ordenação poderá/haverá chaves repetidas pois groupby executa em rdem consecutiva na lista, recriando chaves existentes;

```python
# Q13)
with open("data/segismundo.txt") as f:
    lines = 0
    for line in f:
        lines += 1
    print(lines)
```

```python
with open("data/summary.txt", "w") as g:
    lines = 0
    num_caracteres = 0

    with open("data/segismundo.txt") as f:
        lines = 0
        for line in f:
            lines += 1
        num_caracteres = len(f.read())

    g.write(f"lines={lines}\n")
    g.write(f"chars={num_caracteres}")
```

Q15)
- `FileNotFoundError`: `with open(path) as f:` quando `path` não existe
- `PermissionError`: tentar abrir conexão com um arquivo quando o path está protegido ou sem acesso;
- `FileExistsError`: modo `x` do `open()` exige esxclusividade, esse erro ocorre quando tentamos abrir arquivo que já existe
