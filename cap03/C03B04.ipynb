## Arquivos e o sistema operaciona

É bem simples trabalhar com arquivos no Python, o que constitui um dos motivos pelos quais Python é tão popular para lidar com textos e arquivos.

Para abrir um arquivo para leitura ou escrita, utilize a função embutida `open` com um path de arquivo relativo ou absoluto:

````python
path = "data/segismundo.txt"

f = open(path)
````

Por padrão, o arquivo é aberto em modo somente de leitura 'r'. Podemos então tratar o handle de arquivo `f` como uma lista e iterar pelas linhas, assim:

```python
for line in f:
    pass
```

As linhas são extraídas do arquivo com os marcadores de fim e linha (EOL) intactos, portanto, com frequência, você verá um código para obter uma lista de linhas e um arquivo livres de EOL, da seguinte maneira:

````python
lines = [x.rstrip() for x in open(path)]
lines
````

Quando se usa `open` para criar objetos de arquivos, é importante fechar explicitamente o arquivo ao acabar de utilizá-lo. Fechar o arquivo libera seus recursos, devolvendo-os para o sistema operacional:

````python
f.close()
````

Umas das maneiras de facilitar a limpeza de arquivos abertos é usar a instrução `with`:

````python
with open(path) as f:
    lines = [x.strip() for x in f]
````

Isso fará com que o arquivo `f` seja automaticamente fechado quando saímos do bloco `with`.

Se tivéssemos digitado `f = open(path, 'w')`, um _novo arquivo em data/segismundo.txt_ teria sido criado (tome cuidado!), sobrescrevendo qualquer um em seu lugar. há também o modo de arquivo 'x', que cria um arquivo para escrita, porém falha se o path do arquivo já existir.

Para arquivos de leitura, alguns dos métodos mais comumente utilizados são `read`, `seek` e `tell`. `read` devolve um determinado número de caracteres do arquivo. O que constituium "caractere" é determinado pela codificação do arquivo (por exemplo, UTF-8), ou simplesmente bytes puros se o arquivo for aberto em modo binário:

````python
f = open(path)

f.read(10)
````

````python
f2 = open(path, "rb")  # Modo Binário

f2.read(10)
````

O método `read` faz oposição do handle do arquio avançar de acordo com o número de bytes lidos. `tell` devolve a posição atual:

````python
f.tell()
````

Apesae de termos lidos 10 caracteres do arquivo, a posição é 11 porque foi necessária essa quantidade de bytes para decodificar 10 caracteres usando a codificação default. Podemos verificar a codificação default no módulo `sys`:

````python
sys.getdefaultencoding()
````

`seek` altera a posição do arquivo para byte aí indicado:

````python
f.seek(3)
````

````python
f.read(1)
````

Por fim lembre-se de fechar os arquivos:

````python
f.close()
f2.close()
````

**Modos de arquivo em Python**

| Modo | Descrição |
|------|-----------|
| r | Modo somente de leitura |
| w | Modo somente de escrita; cria um novo arquivo (apagando os dados de qualquer arquivo com o mesmo nome) |
| x | Modo somente de escrita; cria um novo arquivo, mas falha se o path do arquivo já existir |
| a | Cancatena no arquivo existente (cria o arquivo caso ele ainda não exista) |
| r+ | Leitura e escrita |
| b | Adiciona ao modo para arquivos binários (isto é, 'rb' ou 'wb') |
| t | Modo texto para arquivos (decodificando bytes automaticamente para Unicode). É o default se o modo não for especificado. Adicione t a outros modos para utilizar esse modo (isto é, 'rt', 'xt') |

Para escrever texto em um arquivo, use os métodos de arquivo `write` ou `writelines`. Por exemplo, poderíamos criar uma versão de _prof_mod.py_ sem linhas em branco, assim:

````python
with open("data/tmp.txt", "w") as handle:
    handle.writelines(x for x in open(path) if len(x) > 1)
````

````python
with open("data/tmp.txt") as f:
    lines = f.readlines()

lines
````

**Métodos ou atributos de arquivo importantes em Python**

| Método | Descrição |
|--------|-----------|
| read([size]) | Devolve dados do arquivo como uma string, com argumento opcional `size` indicando o número de bytes a ser lido |
| readlines([size]) | Devolve uma lista de linhas do arquivo, com o argumento `size` opcional |
| write(str) | Escreve a string passada em um arquivo |
| writelines(strings) | Escreve a sequência de strings passada no arquivo |
| close() | Fecha o handle |
| flush() | Faz um flush do buffer de E/S interno (descarrega-o) no disco |
| seek(pos) | Move para posição d arquivo indicada (inteiro) |
| tell() | Devolve a posição atual no arquivo como um inteiro |
| closed | True se o arquivo estiver fechado |
