```python
# a)
arr = np.array([6, 7.5, 8, 0, 1])
arr
```

```python
arr.dtype
```

b) Quando criamos um array numérico que contenha int e float, NumPy infere o dtype a float que possui uma maior precisão, assim não havendo perda; Existe uma hierarquia de tipos (bool -> int -> float -> complex -> object);

Quando inferimos o dtype do array para int e existe algum elemento do tipo float, NumPy faz o Cast de float -> int, truncando o mesmo (retirando valores pós .), ele não arredonda;

Q4) `arange` + slicing + atribuição por slice

```python
# a)
arr = np.arange(15)
arr
```

```python
# b)
arr[5:8] = 12
```

```python
# c)
arr
```

Quando atribuimos um valor a um slice (fatia) de um array aquele valor irá se propagar e refletirá no array original, slice são "views" do array original, ou seja, apontam para a refrência inicial;

---

Q5) View vs copy

```python
# a)
a = np.arange(10)

s = a[2:6]
s[:] = -1
```

```python
a
```

```python
s
```

```python
# b)
b = np.arange(10)

t = b[2:6].copy()
t[:] = -1
```

```python
b
```

```python
t
```

c) Quanto reatribuimos `s` com um valor, sendo `s` uma view de `a`, mudamos também os valores referentes em `a`. Já em `b` atribuímos a `t` uma cópia de uma fatia de `b`, onde não existe referência ao array original `b`, sendo assim ao reatribuímos `t` não estamos fazendo nada com `b`;
