## O ndarray do NumPy: um Objeto array multimencional Parte-II

### Indexando com fatias

Assim como os objetos unidimensionais, por exemplo, as listas Python,os ndarrays podem ser fatiados com a sintaxe conhecida:

```python
# imports para estudos
import numpy as np
```

```python
arr = np.array([0, 1, 2, 3, 4, 64, 64, 64, 8, 9])
arr
```

```python
arr[1:6]
```

Considere o array bidimensional `arr2d`, fatiar esse array é um pouco diferente:

```python
arr2d = np.array([[1, 2, 3], [4, 5, 6], [7, 9, 0]])
arr2d
```

```python
arr2d[:2]
```

Com podemos ver, ele foi fatiado ao longo do eixo 0, que é o primeiro eixo. Uma fatia, portanto, seleciona um intervalo de elementos ao longo de um eixo. Ler a expressão `arr2d[:2]` como "selecione as duas primeiras linhas de `arr2d`" talvez possa ajudar.

Podemos passar várias fatias, do mesmo modo como podemos passar vários índices:

```python
arr2d[:2, 1:]
```

Ao fatiar dessa maneira, sempre vamos obter visualizações do array com o mesmo número de dimensões. Ao combinar índices inteiros com fatias, teremos fatias de dimensões menores.

Por exemplo, posso selecionar a segunda linha, mas somente as duas primeiras colunas, assim:

```python
arr2d[1, :2]
```

De modo semelhante, podemos selecionar a terceira coluna, mas somente as duas primeiras linhas, da seguinte forma:

```python
arr2d[:2, 2]
```

É claro que fazer uma atribuição a uma expressão de fatia faz a atribuição para toda a seleção:

```python
arr2d[:2, 1:] = 0
arr2d
```

### Indexação booleana

Vamos considerar um exemplo em que temos alguns dados em um array e um array de nomes com duplicatas.

```python
names = np.array(["Bob", "Joe", "Will", "Bob", "Will", "Joe", "Joe"])
names
```

```python
np.random.seed(1986)
data = np.random.randn(7, 4)
data
```

Suponha que cada nome corresponda a uma linha do array `data` e que queremos selecionar todas as linhas com o nome correspondenete 'Bob'. Assim como nas operações aritméticas, as comparações (como ==) com arrays também são vetorizadas. Assim, comparar `names` com a string 'Bob' produzirá um array _booleano_:

```python
names == "Bob"
```

Esse array booleano pode ser passado quando o array é indexado:

```python
data[names == "Bob"]
```

O array booleano deve ter o mesmo tamanho do eixo do array que ele está indexando. Você pode até mesmo misturar e fazer a correspondência entre arrays booleanos e fatias ou inteiro.

<div style="border:2px solid #ffc107; padding:10px; background-color:#fff3cd; border-radius:5px;">
⚠️ <b>Aviso: A seleção booleana não falhará se o array booleano não tiver o tamanho correto, portanto recomendo tomar cuidado quando utilizar esse recurso.</b> 
</div>
</br>

Nesses exemplos, faremos a seleção a partir das linhas em que `names == "Bob"' e indexo as colunas também:

```python
data[names == "Bob", 2]
```

```python
data[names == "Bob", 3]
```

Para selecionar tudo exceto 'Bob', podemos usar `!=` ou negar a condição uando `~`:

```python
names != "Bob"
```

```python
data[~(names == "Bob")]
```

O operador `~` pode ser útil e quisermos inverter uma condição genérica:

```python
cond = names == "Bob"

data[~cond]
```

Selecionando dois dos três nomes para combinar várias condições booleanas, utilize operadores aritméticos booleanos como `&` (and) e `|` (or):

```python
mask = (names == "Bob") | (names == "Will")
mask
```

```python
data[mask]
```

Selecione dados de um array com indexação booleana _sempre_ criará uma cópia dos dados, mesmo que o array devolvido não seja alterado.

<div style="border:2px solid #ffc107; padding:10px; background-color:#fff3cd; border-radius:5px;">
⚠️ <b>Aviso: As palavras reservadas and e or de Python não funcionam com arrays booleans. Utilize & (and) e | (or) em seu lugar.</b> 
</div>
</br>

Definir valores com arrays booleanos funciona de acordo com o senso comum. Para definir todos os valores negativos em `data` com 0, basta fazer o seguinte:

```python
data[data < 0] = 0
data
```

Definir linhas ou colunas inteiras uando um array booleano unidimensional também é fácil:

```python
data[names != "Joe"] = 7
data
```

Como veremos mais adiante, esses tipos de operações em dados, bidimensionais são covenientes para serem executados com o pandas.

### Indexação sofisticada

_Indexação sofisticada_ (fancy indexing) é o termo adotado pelo NumPy para descrever a indexação usando arrays de inteiros especificando a ordem desejada:

```python
# reatribuição de arr
arr = np.empty((8, 4))
arr
```

```python
for i in range(8):
    arr[i] = i

arr
```

Para selecionar um subconjunto das linhas em uma ordem em particular, podemos simplesmente passar uma lista ou um ndarray de inteiros especificando a ordem desejada:

```python
arr[[4, 3, 0, 6]]
```

```python
arr[[-3, -5, -7]]
```

Passar vários índices de array faz algo um pouco diferente; a instrução seleciona um array unidimensional de elementos correspondentes a cada tupla de índices:

```python
arr = np.arange(32).reshape((8, 4))
arr
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]]
```

Nesse caso, os elementos(1,0),(5,3),(7,1) e (2,2) foram selecionados. Independentemente de quantas dimensões o array tiver, o resultado da indexação sofisticada sempre será unidimensional.

O comportamento da indexação sofisticada nesse caso é um pouco diferente do que alguns usuários poderiam esperar, que é a região retangular formada pela seleção de um subconjunto das linhas e colunas da matriz. Eis uma forma de obter isso:

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]]
```

Tenha em mente que a indexação sofisticada, de modo diferente do fatiamento, sempre copia os dados para um novo array.

### Transposição de arrays e troca de eixos

A transposição é uma forma especial de reformatação que, de modo semelhante, devolve uma visualização dos dados subjacentes, sem copiar nada. Os arrays têm o método `transpose`, além do atributo especial T:

```python
# reatribuição de arr
arr = np.arange(15).reshape(3, 5)
arr
```

```python
arr.T
```

Ao fazer processamento de matrizes ,talvez você realize isto com muita frequência - por exemplo, quando calcular o produto da matriz interna usando `np.dot`:

```python
np.random.seed(1986)
arr1 = np.random.randn(6, 3)
arr1
```

```python
np.dot(arr1.T, arr1)
```

Para arrays de dimensões maiores, `transpose` aceitará uma tupla de números de eixos para permutá-los:

```python
arr2 = np.arange(16).reshape(2, 2, 4)
arr2
```

```python
arr2.transpose(1, 0, 2)
```

Nesse caso, os eixos foram reordenados com o segundo eixo em primeiro lugar, o primeiro eixo em segundo e o último eixo permanexeu inlaterado.

Uma transposição simples com `.T` é um caso especial de troca de eixos. O ndarray tem p método `swapaxes`, que aceita um par de números de eixos e troca os eixos indicados para reorganizar os dados:

```python
arr2.swapaxes(1, 2)
```

De modo semelhante, `swapaxes`devolve uma visualização dos dados, sem criar uma cópia.
