## 3) Correção por questão (acertou / faltou / como corrigir)

### Q01
- Acertou: definiu corretamente **view vs copy** e o efeito de modificar o resultado (propaga vs não propaga).
- Faltou: citar também que **fancy indexing** (índices por lista/array de inteiros) tende a retornar **cópia**.
- Como corrigir: registrar a regra prática completa: **slice → geralmente view**; **boolean/fancy → cópia**.

### Q02
- Acertou: explicou que índice inteiro no eixo (`A[1, :2]`) tende a **reduzir dimensão** (1D) e slice (`A[1:2, :2]`) tende a **preservar dimensão** (2D).
- Faltou: nada relevante.
- Como corrigir: opcional: ancorar com shapes típicos: `(2,)` vs `(1,2)`.

### Q03
- Acertou: percebeu que pode ocorrer **erro** quando não há compatibilidade.
- Faltou/Impreciso: em **indexação booleana**, a regra é **compatibilidade com o eixo indexado** (ex.: filtrar linhas exige máscara do tamanho do nº de linhas). “Broadcasting automático” aqui não é o comportamento esperado.
- Como corrigir: afirmar direto: máscara incompatível com o eixo → **erro**; se quiser usar broadcasting, precisa **ajustar shape explicitamente** para ficar compatível com o alvo.

### Q04
- Acertou (parcial): identificou que ambos podem retornar 2D.
- Erro central: a diferença não é “1 coluna vs 2 colunas”.
  - `A[[1,5,7,2]]` = seleção de **linhas** (mantém todas as colunas) → **2D**.
  - `A[[1,5,7,2], [0,3,1,2]]` = seleção de **pares (linha,coluna)** ponto a ponto → **1D**.
- Como corrigir: memorizar: **dois vetores na mesma indexação = coordenadas pareadas**, não submatriz.

### Q05
- Acertou: identificou corretamente que deve usar `|` (ou `&`) em vez de `or`/`and`.
- Faltou: expressão correta com **parênteses por condição** e posicionamento do filtro no array.
- Como corrigir: forma segura: `(names == "Bob") | (names == "Will")` (cada comparação entre parênteses) e aplicar isso no eixo correto do array (linhas).

### Q06
- Acertou: `arr[::2]` (step 2) e `arr[::-1]` (inversão); ambos como slicing, então **tendem a view**.
- Faltou: nada relevante.
- Como corrigir: opcional: mencionar que slices reconfiguram *strides* (sem copiar dados).

### Q07
- Acertou: `data[mask] = 0` altera **o original**; `filtered = data[mask]` cria **cópia** e zerar `filtered` não volta para `data`.
- Faltou: nada relevante.
- Como corrigir: manter como regra: “atribuição no alvo original” vs “mutação em cópia”.

### Q08
- Acertou: transposição (`A.T`) troca eixos em 2D e é usada para ajustar orientação em operações matriciais.
- Faltou: regra correta de compatibilidade no produto matricial: **colunas do 1º = linhas do 2º**.
- Como corrigir: afirmar sem ambiguidade: se `A` é `(m,n)` e `B` é `(n,p)`, então `A @ B` é `(m,p)`; `A.T` vira `(n,m)`.

### Q09
- Acertou: citou “permutação de eixos”.
- Faltou: exemplo concreto + efeito no `shape`; a ideia de “um eixo permanece de ponte” não define a operação.
- Como corrigir: dar exemplo explícito: shape `(2,2,4)` com permutação `(2,0,1)` → `(4,2,2)` (reordenação de dimensões).

### Q10
- Acertou (mínimo): ambos se relacionam a rearranjo de eixos.
- Faltou/Confuso: definição operacional.
  - `swapaxes(i,j)` troca **somente dois eixos**.
  - `transpose(permutação)` faz **permutação geral** de eixos (em 2D, equivale a trocar 0↔1).
- Como corrigir: regra prática: “troca pontual de 2 eixos” → `swapaxes`; “reordenar vários eixos” → `transpose`.

### Q11
- Acertou: citou precedência/hierarquia de operações e necessidade de parênteses.
- Faltou: conectar explicitamente ao NumPy: com `&` e `|`, sem parênteses, a expressão pode ser interpretada em ordem errada e quebrar comparações.
- Como corrigir: padrão obrigatório: `(cond1) & (cond2)` e `(cond1) | (cond2)`.

### Q12
- Acertou: `A[rows, cols]` retorna “pontos” por **pares (row_i, col_i)**.
- Faltou: explicar como obter a **submatriz retangular** (produto cartesiano rows×cols).
- Como corrigir: abordagem correta é construir seleção “malhada” (rows×cols) ou fazer em duas etapas: selecionar linhas e depois colunas (conceitualmente: primeiro “rows”, depois “cols”).

## 4) Checklist final (3–7 itens)
1) Indexação booleana: máscara **compatível com o eixo** (incompatível → erro).  
2) Fancy indexing 2D: diferença entre **seleção de linhas** vs **pares (i,j)**.  
3) Como obter submatriz retangular com listas de índices (rows×cols, não pares).  
4) `transpose` em nD: permutação de eixos com exemplo e impacto no `shape`.  
5) `swapaxes` vs `transpose`: troca de 2 eixos vs permutação geral.  
6) Expressões booleanas: sempre usar **parênteses** com `&` e `|`.
