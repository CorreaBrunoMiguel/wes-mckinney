## 3) Correção por questão

### Q1 (12/12)
Correto.
- `x = [10, -1, 30, -1, 50, 60]`
- `y = [-1, -1]`
- `y` é **view** (slice), então altera `x`.

### Q2 (8/8)
Correto.
- `A[1:4]` → tende a **view**, mutação propaga.
- `A[A > 0]` → tende a **cópia**, mutação não propaga.

### Q3 (7/10)
1) Qtde de `True`: **4** (correto)  
2) Shape de `A[mask]`: você colocou “3 x 3” (incorreto).  
   - Correto: **(4,)** (1D com os selecionados)  
3) Valores: **[2, 5, 3, 4]** (correto, ordem row-major)

### Q4 (8/8)
Correto (explicação + linha correta).
- `sel = (v >= 3) & (v <= 7)`
- Motivo: `and` tenta reduzir o array a um booleano único; em NumPy usa-se `&`/`|` com parênteses.

### Q5 (8/14)
Enunciado pedia **valor e shape** + explicação.
1) `M[[2, 0]]`  
   - Valor: você acertou `[[9,10,11,12],[1,2,3,4]]`  
   - Shape: você colocou **(2,2)** (incorreto). Correto: **(2,4)**  
2) `M[[2, 0], [1, 3]]`  
   - Valor correto: **[10, 4]** (você chegou nos valores, apesar da forma “coluna” no texto)  
   - Shape correto: **(2,)**  
3) Por que não é submatriz retangular?  
   - Você respondeu “Não sei” (faltou).  
   - Correto: dois vetores na mesma indexação são tratados como **pares (linha,coluna)** (zip), não como produto cartesiano.

### Q6 (4/10)
1) Por que `M[rows, cols]` não resolve: você acertou a ideia (“gera pontos/pares ordenados”).  
2) Forma correta: faltou (ex.: `M[np.ix_(rows, cols)]` ou `M[rows][:, cols]`).

### Q7 (6/8)
1) `X.transpose(2,0,1)` → **(4,2,3)** (correto)  
2) `X.swapaxes(0,2)` → **(4,3,2)** (correto)  
3) Diferença de intenção: faltou.  
   - Correto: `swapaxes` troca **dois eixos**; `transpose` faz **permutação geral**.

### Q8 (9/10)
1) `A @ B` válido, shape correto é **(5,7)** (você colocou **(5,5)**).  
2) `B @ A` inválido (correto)  
3) `A.T @ C` válido, shape **(2,2)** (correto)  
4) `A @ C` inválido (correto)

### Q9 (7/10)
1) Explicação do bug: correta (máscara precisa casar com nº de linhas).  
2) Código: sua máscara funciona, mas ficou **hardcoded**. O pedido era “construir a máscara a partir de `id_`”.  
   - Esperado: `mask = (id_ == "X")` e então `D[mask] = 0`.

### Q10 (4/10)
1) View vs cópia: correto (slice reverso tende a **view**)  
2) Contiguidade: você disse que compartilha memória, mas não respondeu “contíguo sim/não e por quê”.  
   - Correto: frequentemente **não é contíguo** (strides negativos).  
3) Situação para forçar cópia: faltou (ex.: garantir contiguidade/performance: `r = x[::-1].copy()`).
