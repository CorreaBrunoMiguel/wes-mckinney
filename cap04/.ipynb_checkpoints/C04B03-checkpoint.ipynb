## Funções universais: funções rápidas de arrays para todos os elementos

Uma função universal, ou _ufunc_, é uma função que executa operaçoes em todos os elementos nos dados de ndarrays. Podemos pensar nelas como _wrappers_ vetorizados rápidos para funções simples que aceitam um ou mais valores escalares e geram um ou mais resultados escalares.

Muitas funções são transformações simples em todos os elementos, como _sqrt_ ou _exp_:

```python
# imports
import numpy as np
```

```python
arr = np.arange(10)
arr
```

```python
np.sqrt(arr)
```

```python
np.exp(arr)
```

Essasfunções são chamadas de _ufuncs unárias_. Outras, como `add` ou `maximun`, aceitam dois arrays e devolvem um único array como resultado:

```python
np.random.seed(1986)
x = np.random.randn(8)
x
```

```python
np.random.seed(2026)
y = np.random.randn(8)
y
```

```python
np.maximum(x, y)
```

Neste caso, `numpy.maximum` calculou o máximo entre os elementos de x e y para todos os elementos.

Embora não seja comum, uma ufunc pode devolver vários arrays. `modf` é um exemplo; ela é uma versão vetorizada da função embutida `divmod` de Python e devolve as partes fracionárias e inteira de um array de ponto flutuante:

```python
np.random.seed(1990)
arr1 = np.random.randn(7) * 5
arr1
```

```python
remainder, whole_part = np.modf(arr1)

remainder
```

```python
whole_part
```

As ufuncs aceitam um argumento opcinal _out_ que lhes permite atuar _in-place_ns arrays:

```python
arr1
```

```python
np.sqrt(arr1)
```

```python
np.sqrt(arr1, arr1)
```

```python
arr1
```

**Ufuncs unárias**
| Função | Descrição |
|--------|-----------|
| abs, fabs | Calcula o valor absoluto de inteiros, números de ponto flutuante e valores cmplexos para todos os elementos |
| sqrt | Calcula a raiz quadrada de cada elemento |
| square | Calcula o quadrado de cada elemento |
| exp | Calcula o exponencial $e^x$ de cada elemento |
| log, log10, log2, log1p | Logaritmo natural, na base 10, na base 2 e ($1 + x$), respectivamente |
| sign | Calcula o sinal de cada elemento: 1 (positivo), 0 (zero) ou -1(negativo) |
| ceil | Calcula o teto de cada elemento |
| floor | Calcula o piso de cada elemento |
| rint | Arredonda os elementos para o inteiro mais próximo, preservando o _dtype_ |
| modf | Devolve as partes fracinárias e inteira do array como um array separado |
| isnan | Devolve um array booleano indicando se cada valor é __NaN__ |
| isfinite, isinf | Devolve um array booleano indicando se cada elemento é finito ou infinito, respectivamente |
| cos, cosh, sin, sinh, tan, tanh | Funções trigonométricas regulares e hiperbólicas |
| arccos, arccosh, arcsin, arcsinh, arctan, arctanh | Funções trigonométricas inversas |
| logical_not | Calcula o valor-verdade de _not x_ para todos os elementos |

**Funções universais binárias**
| Função | Descrição |
|--------|-----------|
| add | Soma elementos correspondentes em arrays |
| subtract | Subtrai elementos do segundo array do primeiro |
| multiply | Multiplica elementos do array |
| divide, floor_divide | Faz a divisão ou a divisão pelo piso (truncando o resto) |
| power | Elevaos elementos do primeiro array às potências indicadas no segundo array |
| maximun, fmax | Máximo para todos os elementos; __fmax__ ignora __NaN__ |
| minimum, fmin | Mínimo para todos os elementos; __fmin__ ignora __NaN__ |
| mod | Módulo para todos os elementos (resto da divisão) |
| copysign | Copia o sinal dos valores do segundo argumento para os valores do primeiro argumento |
| greater, greater_equal, less, less_equal, equal, not_equal | Faz uma comparação para todos os elementos, produzindo um array booleano |
|logical_and, logical_or, logical_xor | Calcula o valor-verdade da operação lógica para todos os elementos |
