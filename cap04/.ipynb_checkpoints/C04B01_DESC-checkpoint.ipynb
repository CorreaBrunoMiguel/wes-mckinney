## C04B01 - Respostas Descritivas

1. É um objeto N-dimensional, um contêiner rápido e flexívelmpara conjuntos de dados grandes. Ele é homgêneo, isto é, seus elementos devem possuir o mesmo tipo (dtype). Com um ndarray podemos realizar operações matemáticas em blocos inteiros com maior rapidez e menor custo de memória;

2. `shape` -> tupla que indica o tamanho de cada dimensão; `ndim` -> diz qual a dimensão do array; `dtype` -> retorna o tipo de dados do array.

Ex: temos o array `arr = np.array([[1, 2, 3], [4, 5, 6]])`

`arr.shape` -> (2, 3)
`arr.ndim` -> 2
`arr.dtype` -> dtype('int64')

3. a) objetivo:
- `np.array`: gerar um ndarray de um sequência, (ou lista de listas de mesmo tamanho)
- `np.zeros`: gerar um ndarray de 0s, podendo escolher shape, dtypes
- `np.empty`: gerar um ndarray "vazio" ou com "lixos"
- `np.arange`: gerar um ndarray de elementos consecutivos, podendo escolher inicio e fim

b) quando usar
- `np.array`: quando queremos transformar sequências homegêneas em vetor
- `np.zeros`: ter ndarray com zeros
- `np.empty`: precisamos de um ndarray "vazio"
- `np.arange`: queremos um array com elementos consecutivos

c) risco/pegadinha
- `np.array`: quando temo lista de listas com tamanhos diferentes
- `np.zeros`: multiplicar um array por np.zeros
- `np.empty`: gera lixo em muitos casos, prefira zeros
- `np.arange`: valor final é inclusivo

4. Ele irá determinar float como dtype, pois NumPy faz o cast para tipo mais acessível.

5. `astype` ele cria um novo array baseado no array de origem; quando precisamos garantir o dtype do array temos que usar astype;

6. Ele trunca o valor (retira a parte decimal), importante saber que ele não faz arredondamento de nenhum tipo;

Ex: temos `array([2.3, 4.78, -1.345])` se fizermos `astype(np.int32)` deste array teremos `array([2, 4, -1])`

7. Acredito pelo fato de estarmos usando vetores ao invés de sequências que necessitam de um laço for para percorrer os dados. Quando usamos vetores NumPy, Python faz os cálculos de uma vez;

8. `arr[5:8]` irá fatiar aquele intervalo, criará uma visualização do array de origem apontando para apenas uma memória (do array original `arr`); se fizermos `arr[5:8] = 12` os valores deste intervalo em `arr` tornarão toos 12;

9. `view` é uma visualização do array original, uma referência a memória (localização) do array original. Para evitar o `view` podemos usar `.copy()` que copia o array;

10. Os dois retornam o mesmo "valor", mas `arr2d[0, 2]` é preferível pois retorna diretamente sem ter que passar por mais de uma "geração", acessando recursivamente;

11. `arr3d[0]` -> retorna um array 2D, visualmente seria a primeira fatia do cubo (acima); `arr3d[1, 0]` -> um array unidimensional, algo como a segunda fatia gera um array 2d aí pegamos o primeiro array;

12. Quando queremos garantir a integridade original do array; quando precismos trabalhar em pedaços e necessitamos a garantia de um array para comparar?
